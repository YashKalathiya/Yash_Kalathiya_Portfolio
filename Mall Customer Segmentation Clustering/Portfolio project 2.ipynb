# Clustering Univariate, Bivariate, and Multivariate

```python
clustering1 = KMeans(n_clusters=3)
```

```python
clustering1.fit(df[['Annual Income (k$)']])
```

```python
clustering1.labels_
```

```python
df['Income Cluster'] = clustering1.labels_
df.head()
```

```python
df['Income Cluster'].value_counts()
```

```python
clustering1.inertia_
```

```python
inertia_scores = []
for i in range(1,11):
    kmeans = KMeans(n_clusters = i)
    kmeans.fit(df[['Annual Income (k$)']])
    inertia_scores.append(kmeans.inertia_)
```

```python
inertia_scores
```

```python
plt.plot(range(1,11), inertia_scores)
```

```python
df.groupby('Income Cluster')[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].mean()
```

```python
# Bivariate Clustering
```

```python
clustering2 = KMeans(n_clusters = 5)
clustering2.fit(df[['Annual Income (k$)','Spending Score (1-100)']])
clustering2.labels_
df['Spending And Income Cluster'] = clustering2.labels_
df.head()
```

```python
inertia_scores2 = []
for i in range(1,11):
    kmeans2 = KMeans(n_clusters = i)
    kmeans2.fit(df[['Annual Income (k$)', 'Spending Score (1-100)']])
    inertia_scores2.append(kmeans2.inertia_)
plt.plot(range(1,11), inertia_scores2)
```

```python
centersdf = pd.DataFrame(clustering2.cluster_centers_)
centersdf.columns = ['x','y']
centersdf
```

```python
plt.figure(figsize=(10,8))
plt.scatter(x=centersdf['x'], y = centersdf['y'], s=100, c='black',marker='*')
sns.scatterplot(data = df, x = 'Annual Income (k$)', y = 'Spending Score (1-100)', hue = 'Spending And Income Cluster', palette = 'tab10')
plt.savefig('Clustering_bivariate.png')
```

```python
pd.crosstab(df['Spending And Income Cluster'], df['Gender'], normalize = 'index')
```

```python
df.groupby('Spending And Income Cluster')[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].mean()
```

```python
# Multivariate Cluster
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scale = StandardScaler()
```

```python
df.head()
```

```python
df2 = pd.get_dummies(df,drop_first=True)
df2.head()
```

```python
df2.columns
```

```python
df2 = df2 [['Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Gender_Male']]
df2.head()
```

```python
df2 = pd.DataFrame(scale.fit_transform(df2))
```

```python
inertia_scores3 = []
for i in range(1,11):
    kmeans3 = KMeans(n_clusters = i)
    kmeans3.fit(df2)
    inertia_scores3.append(kmeans3.inertia_)
plt.plot(range(1,11), inertia_scores3)
```
